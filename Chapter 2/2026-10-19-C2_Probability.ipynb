```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import expon, norm, lognorm

# Set up 6 panel figure (3 rows, 2 columns)
fig, axs = plt.subplots(3, 2, figsize=(10, 15))

# 1. Exponential Distribution (Top Row)
xdens_exp = np.arange(0, 5.02, 0.02)

# Upper Left: Exponential p.d.f.
# Note: scipy's scale parameter is 1/lambda
axs[0, 0].plot(xdens_exp, expon.pdf(xdens_exp, scale=1/0.5), '-', label=r'$\lambda=0.5$')
axs[0, 0].plot(xdens_exp, expon.pdf(xdens_exp, scale=1/1.0), '--', label=r'$\lambda=1.0$')
axs[0, 0].plot(xdens_exp, expon.pdf(xdens_exp, scale=1/1.5), ':', label=r'$\lambda=1.5$')
axs[0, 0].set_ylim(0, 1.5)
axs[0, 0].set_ylabel('Exponential p.d.f.')
axs[0, 0].legend(frameon=False)

# Upper Right: Exponential c.d.f.
axs[0, 1].plot(xdens_exp, expon.cdf(xdens_exp, scale=1/0.5), '-', label=r'$\lambda=0.5$')
axs[0, 1].plot(xdens_exp, expon.cdf(xdens_exp, scale=1/1.0), '--', label=r'$\lambda=1.0$')
axs[0, 1].plot(xdens_exp, expon.cdf(xdens_exp, scale=1/1.5), ':', label=r'$\lambda=1.5$')
axs[0, 1].set_ylim(0, 1.0)
axs[0, 1].set_ylabel('Exponential c.d.f.')
axs[0, 1].legend(loc='center right', frameon=False)

# 2. Normal (Gaussian) Distribution (Middle Row)
xdens_norm = np.arange(-5, 5.02, 0.02)

# Middle Left: Normal p.d.f.
axs[1, 0].plot(xdens_norm, norm.pdf(xdens_norm, loc=0, scale=np.sqrt(0.2)), '-', label=r'$\mu=0, \sigma^2=0.2$')
axs[1, 0].plot(xdens_norm, norm.pdf(xdens_norm, loc=0, scale=np.sqrt(1.0)), '--', label=r'$\mu=0, \sigma^2=1.0$')
axs[1, 0].plot(xdens_norm, norm.pdf(xdens_norm, loc=0, scale=np.sqrt(5.0)), ':', label=r'$\mu=0, \sigma^2=5.0$')
axs[1, 0].plot(xdens_norm, norm.pdf(xdens_norm, loc=-2.0, scale=np.sqrt(0.5)), '-.', label=r'$\mu=-2, \sigma^2=0.5$')
axs[1, 0].set_ylim(0, 1.0)
axs[1, 0].set_ylabel(r'$\phi_{\mu \sigma^2}(x)$')
axs[1, 0].legend(frameon=False)

# Middle Right: Normal c.d.f.
axs[1, 1].plot(xdens_norm, norm.cdf(xdens_norm, loc=0, scale=np.sqrt(0.2)), '-', label=r'$\mu=0, \sigma^2=0.2$')
axs[1, 1].plot(xdens_norm, norm.cdf(xdens_norm, loc=0, scale=np.sqrt(1.0)), '--', label=r'$\mu=0, \sigma^2=1.0$')
axs[1, 1].plot(xdens_norm, norm.cdf(xdens_norm, loc=0, scale=np.sqrt(5.0)), ':', label=r'$\mu=0, \sigma^2=5.0$')
axs[1, 1].plot(xdens_norm, norm.cdf(xdens_norm, loc=-2.0, scale=np.sqrt(0.5)), '-.', label=r'$\mu=-2, \sigma^2=0.5$')
axs[1, 1].set_ylim(0, 1.0)
axs[1, 1].set_ylabel(r'$\Phi_{\mu \sigma^2}(x)$')
axs[1, 1].legend(loc='lower right', frameon=False)

# 3. Lognormal Distribution (Bottom Row)
xdens_lnorm = np.arange(0.01, 3.02, 0.02) # Start slightly above 0 for lognormal

# Bottom Left: Lognormal p.d.f.
# scipy's lognorm parameters: s = sdlog, scale = exp(meanlog)
axs[2, 0].plot(xdens_lnorm, lognorm.pdf(xdens_lnorm, s=5, scale=np.exp(0)), '-', label=r'$\sigma=5$')
axs[2, 0].plot(xdens_lnorm, lognorm.pdf(xdens_lnorm, s=1, scale=np.exp(0)), '--', label=r'$\sigma=1$')
axs[2, 0].plot(xdens_lnorm, lognorm.pdf(xdens_lnorm, s=0.5, scale=np.exp(0)), ':', label=r'$\sigma=1/2$')
axs[2, 0].plot(xdens_lnorm, lognorm.pdf(xdens_lnorm, s=0.125, scale=np.exp(0)), '-.', label=r'$\sigma=1/8$')
axs[2, 0].set_ylim(0, 2)
axs[2, 0].set_ylabel('Lognormal density')
axs[2, 0].legend(loc='upper right', frameon=False)

# Bottom Right: Lognormal c.d.f.
axs[2, 1].plot(xdens_lnorm, lognorm.cdf(xdens_lnorm, s=5, scale=np.exp(0)), '-', label=r'$\sigma=5$')
axs[2, 1].plot(xdens_lnorm, lognorm.cdf(xdens_lnorm, s=1, scale=np.exp(0)), '--', label=r'$\sigma=1$')
axs[2, 1].plot(xdens_lnorm, lognorm.cdf(xdens_lnorm, s=0.5, scale=np.exp(0)), ':', label=r'$\sigma=1/2$')
axs[2, 1].plot(xdens_lnorm, lognorm.cdf(xdens_lnorm, s=0.125, scale=np.exp(0)), '-.', label=r'$\sigma=1/8$')
axs[2, 1].set_ylim(0, 1.0)
axs[2, 1].set_xlabel('x')
axs[2, 1].set_ylabel('Lognormal distribution')
axs[2, 1].legend(loc='center right', frameon=False)

plt.tight_layout()
plt.show()
```
